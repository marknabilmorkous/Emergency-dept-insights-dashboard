```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/sample_data/healthcare  dataset.csv')
```

```python
print(df.isnull().sum())
```

```python
df['High_Risk_Readmit'] = np.where((df['Age'] >= 65) & (df['Readmitted_30D'] == 'Yes'), 1, 0)
df['Wait_Category'] = pd.cut(df['Wait_Time_Min'],
                             bins=[-1, 15, 60, 120, 1000],
                             labels=['Immediate (<15m)', 'Moderate (15-60m)', 'Long (60-120m)', 'Critical (>120m)'])
```

```python
summary_triage = df.groupby('Triage_Level').agg(
    Total_Patients=('Patient_ID', 'count'),
    Avg_Wait_Min=('Wait_Time_Min', 'mean'),
    Avg_Stay_Hrs=('Length_Of_Stay_Hrs', 'mean'),
    Total_Cost=('Total_Cost_USD', 'sum')
).reset_index()
```

```python
print(summary_triage)
```

```python
df.to_csv('cleaned_healthcare_er.csv', index=False)
```
